# Marbella price data: quick tour

Three open datasets from [Marbella Wire](https://marbellawire.com): the Sunbed Index,
the chiringuito census and the beach club price census. All prices in EUR,
read from each venue's own published menu, rate card or booking system.
License: [CC BY 4.0](https://creativecommons.org/licenses/by/4.0/).

The cells below load the CSVs straight from the repo, so this notebook is
always as current as the repo's latest refresh.

```python
import pandas as pd

sunbed = pd.read_csv('https://raw.githubusercontent.com/shiftdylson1/marbella-price-data/main/sunbed-index/sunbed-index.csv')
chiringuitos = pd.read_csv('https://raw.githubusercontent.com/shiftdylson1/marbella-price-data/main/chiringuitos/chiringuitos.csv')
beach_clubs = pd.read_csv('https://raw.githubusercontent.com/shiftdylson1/marbella-price-data/main/beach-clubs/beach-clubs.csv')
print(f'{len(sunbed)} sunbed index rows, {len(chiringuitos)} chiringuitos, {len(beach_clubs)} beach clubs')
```

## Beach club sunbed prices

Cheapest sunbed per club, high season, per day. The dashed line is the
Sunbed Index average (two sunbeds plus minimum spend, so a different
basis: it is context, not the mean of these bars).

```python
import matplotlib.pyplot as plt

priced = beach_clubs.dropna(subset=['sunbed_price']).sort_values('sunbed_price')
index_avg = sunbed['index_average_eur'].iloc[0]

fig, ax = plt.subplots(figsize=(10, max(4, len(priced) * 0.3)))
ax.barh(priced['name'], priced['sunbed_price'], color='#2a6f97')
ax.axvline(index_avg, linestyle='--', color='#d1495b',
           label=f'Sunbed Index average: {index_avg:.0f} EUR')
ax.set_xlabel('EUR per day (high season)')
ax.set_title(f"Beach club sunbed prices, season {int(beach_clubs['season_year'].iloc[0])}")
ax.legend()
plt.tight_layout()
plt.show()
```
